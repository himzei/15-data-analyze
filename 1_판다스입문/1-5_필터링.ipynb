```python
mask1 = (titanic.age >= 10) & (titanic.age < 20 )
df_teenage = titanic[mask1]
df_teenage.head()
```

```python
mask2 = (titanic.age < 10) & (titanic.sex == "female")
df_female_under10 = titanic.loc[mask2]
df_female_under10.shape
```

```python
mask3 = (titanic.age < 10) | (titanic.age >= 60) 
mask3
```

```python
df_under10_morethan60 = titanic.loc[mask3, ["age", "sex", "alone"]]
df_under10_morethan60
```

```python
## query() 매서드 활용

import pandas as pd
import seaborn as sns 

titanic = sns.load_dataset("titanic")
df = titanic.loc[ : , ["age", "fare"]]
df
```

```python
## 20세 미만의 조건을 만족하는 행을 필터링
df.query("age < 20")
```

```python
df.query("age >= 20 or fare < 15")
```

```python
## isin() 매서드 활용

## DF객체.isin(추출 값의 리스트)

import pandas as pd
import seaborn as sns 

titanic = sns.load_dataset("titanic")

pd.set_option("display.max_columns", 10)

titanic.head()
```

```python
mask1 = titanic["embark_town"] == "Southhampton"
mask2 = titanic["embark_town"] == "Queenstown"

df_boolean = titanic[mask1 | mask2]

df_boolean
```

```python
isin_filter = titanic["embark_town"].isin(["Southhampton", "Queenstown"])
df_isin = titanic[isin_filter]
df_isin
```
