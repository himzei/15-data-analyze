```python
## 시리즈 vs 시리즈 

dict_data1 = {
    "국어": 100, 
    "영어": 80, 
    "수학": 90
}

dict_data2 = {
    "수학": 80,
    "국어": 90, 
    "영어": 80, 
}

student1 = pd.Series(dict_data1)
student2 = pd.Series(dict_data2)

print(student1)
print()
print(student2)
```

```python
addition = student1 + student2
print(type(addition))
addition
```

```python
substraction = student1 - student2
multiplication = student1 * student2 
division = student1 / student2 

print(addition)
print(substraction)
print(multiplication)
print(division)

```

```python
lis = [
    addition,
    substraction,
    multiplication, 
    division
]

result = pd.DataFrame(lis, index=["덧셈", "뺄셈", "곱셈", "나눗셈"])
result
```

```python
## NaN 값이 있는 시리즈 연산

import pandas as pd 
import numpy as np 

dict_data1 = {
    "국어": np.nan, 
    "영어": 80, 
    "수학": 90
}

dict_data2 = {
    "수학": 80,
    "국어": 90, 
    
}

student1 = pd.Series(dict_data1)
student2 = pd.Series(dict_data2)

print(student1)
print()
print(student2)
```

```python
addition = student1 + student2
addition
```

```python
substraction = student1 - student2 
multiplation = student1 * student2
division = student1 / student2

result = pd.DataFrame([addition, substraction, multiplation, division], index=["덧셈", "뺄셈", "곱셈", "나눗셈"])
result 
```

```python
## 연산메소드 
import pandas as pd 
import numpy as np 

dict_data1 = {
    "국어": np.nan, 
    "영어": 80, 
    "수학": 90
}

dict_data2 = {
    "수학": 80,
    "국어": 90, 
    
}

student1 = pd.Series(dict_data1)
student2 = pd.Series(dict_data2)

sr_add = student1.add(student2, fill_value=0)
sr_sub = student1.sub(student2, fill_value=0)
sr_mul = student1.mul(student2, fill_value=0)
sr_div = student1.div(student2, fill_value=0)

lis = [
    sr_add,
    sr_sub,
    sr_mul,
    sr_div
]

result = pd.DataFrame(lis, index=["덧셈", "뺄셈", "나눗셈", "곱셈"])
result 
```
