```python
# csv 파일로 저장


import pandas as pd

data = {
    "name": ["jerry", "Riah", "Paul"], 
    "algol": ["A*", "A+", "B"], 
    "basic": ["C", "B", "B+"], 
    "c++": ["B+", "C", "C+"]
}

df = pd.DataFrame(data)
df =df.set_index("name")
df
```
