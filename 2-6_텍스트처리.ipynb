```python
## 문자열 매서드

import pandas as pd
import numpy as np 


list_data = [
    "Apple_사과", 
    "Banana_바나나", 
    "Cherry_체리",
    np.nan
]

ser = pd.Series(
    list_data, 
    index=["First ", " Second", " Third", "Fourth"], 
    dtype="string"
)

ser
```

```python
## 소문자로 변형 -> str접근자.lower()
print(ser.str.lower())

## 대문자로 변형 -> str접근자.upper()
print(ser.str.upper())

## 문자열 길이 계산 -> str접근자.len()
# len(ser) -> 내장함수 len() : 객체의 행의수를 구할때 
print(ser.str.len())
```

```python
## 문자열 분할
exam1 = ser.str.split("_")
print(exam1)
type(exam1)
```

```python
## 문자열으르 분할하여 데이터프레임으로 확장
exam2 = ser.str.split("_", expand=True)
print(exam2)
print(type(exam2))
```

```python
## 문자열 중 특정 요소 선택
ser.str.split("_").str.get(1)
```

```python
## 행 인덱스 추출

idx = ser.index
print(idx)
```
